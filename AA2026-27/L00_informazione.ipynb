## Il gioco di Shannon

## Formalizzare l'incertezza: l'entropia

Nel primo gioco, con N numeri ugualmente possibili, servivano log₂(N) domande in media per individuare quello giusto (l'abbiamo verificato con la strategia a bisezione). Questa quantità ha un nome: si chiama **entropia**, ed è la formalizzazione dell'idea intuitiva di incertezza con cui abbiamo giocato finora.

Quando tutti gli esiti sono ugualmente probabili, l'entropia è semplicemente:

$$H = \log_2(N)$$

Verifichiamo che torni con quello che abbiamo osservato.

```python
from nlp.entropy import equiprobable_entropy

for n in (2, 4, 8, 16, 100):
    print(f"N={n}: H = {equiprobable_entropy(n):.2f} bit")
```

## Il caso di un testo casuale

Proviamo ad applicare la stessa formula a un testo. Se usiamo un alfabeto di 26 lettere più lo spazio (27 simboli), e ogni simbolo è ugualmente probabile, l'entropia per carattere è log₂(27).

Generiamo un testo casuale con queste caratteristiche e verifichiamo che l'entropia calcolata sulle frequenze osservate corrisponda a quella teorica.

```python
from nlp.entropy import equiprobable_entropy, random_text, letter_frequencies, entropy

ALPHABET = "abcdefghijklmnopqrstuvwxyz "
random_sample = random_text(500, ALPHABET)

print(random_sample)
print(f"\nEntropia teorica (27 simboli equiprobabili): {equiprobable_entropy(len(ALPHABET)):.2f} bit")
print(f"Entropia calcolata sulle frequenze osservate: {entropy(letter_frequencies(random_sample)):.2f} bit")
```

## E per un testo vero?

Nel gioco di Shannon abbiamo visto che le lettere di una parola reale non sono tutte ugualmente prevedibili: alcune si indovinano subito, altre no. Questo succede perché in una lingua reale le lettere non compaiono con la stessa frequenza — pensiamo a quante parole italiane contengono la "e" rispetto a quante contengono la "q".

Quando le probabilità non sono tutte uguali, la formula dell'entropia si generalizza:

$$H = -\sum_i p_i \log_2(p_i)$$

dove $p_i$ è la probabilità di ciascun simbolo. Quando tutte le probabilità sono uguali, questa formula coincide esattamente con log₂(N) di prima.

Calcoliamola su un breve testo italiano reale, e confrontiamola con quella del testo casuale di prima.

```python
from nlp.entropy import normalize_text

real_text = normalize_text("""
Quel ramo del lago di Como, che volge a mezzogiorno, tra due catene non
interrotte di monti, tutto a seni e a golfi, a seconda dello sporgere e del
rientrare di quelli, vien, quasi a un tratto, a ristringersi, e a prender
corso e figura di fiume, tra un promontorio a destra, e un'ampia costiera
dall'altra parte.
""")

print(f"Entropia del testo reale: {entropy(letter_frequencies(real_text)):.2f} bit")
print(f"Entropia del testo casuale: {entropy(letter_frequencies(random_sample)):.2f} bit")
print(f"Entropia massima possibile (27 simboli equiprobabili): {equiprobable_entropy(27):.2f} bit")
```

## L'informazione come riduzione di incertezza

Se l'entropia misura quanta incertezza c'è su qualcosa, allora sapere qualcosa in più dovrebbe farla diminuire. La differenza tra l'incertezza iniziale e quella rimasta dopo aver saputo qualcosa è esattamente la quantità di informazione che quella conoscenza ci ha fornito:

$$I = H_{prima} - H_{dopo}$$

Proviamo a calcolarla su un paio di esempi concreti, tratti da giochi che conosciamo già.

```python
from nlp.entropy import equiprobable_entropy

h_carta = equiprobable_entropy(52)
h_carta_dato_seme = equiprobable_entropy(13)
print(f"Incertezza su quale carta è (mazzo da 52): {h_carta:.2f} bit")
print(f"Incertezza rimasta, sapendo il seme (13 carte per seme): {h_carta_dato_seme:.2f} bit")
print(f"Informazione fornita dal sapere il seme: {h_carta - h_carta_dato_seme:.2f} bit")

print()

h_dado = equiprobable_entropy(6)
h_dado_dato_parita = equiprobable_entropy(3)
print(f"Incertezza sul risultato di un dado: {h_dado:.2f} bit")
print(f"Incertezza rimasta, sapendo se è pari o dispari: {h_dado_dato_parita:.2f} bit")
print(f"Informazione fornita dal sapere la parità: {h_dado - h_dado_dato_parita:.2f} bit")
```
